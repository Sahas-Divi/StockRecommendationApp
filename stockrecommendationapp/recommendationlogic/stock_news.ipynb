```python
from langchain.agents import create_agent
import os
from dotenv import load_dotenv
import getpass
from langchain_tavily import TavilySearch
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI


load_dotenv()
openai_api_key = os.getenv("OPENAI_API_KEY")
tavily_api_key = os.getenv("TAVILY_API_KEY")

tavily_search_news_tool = TavilySearch(
    max_results=5,
    topic="news",
    time_range="week",
)

llm = ChatOpenAI(
    model="gpt-5-mini",
)

news_agent = create_agent(
    model=llm,
    tools=[tavily_search_news_tool],
    system_prompt="You are a helpful assistant",
)

result = news_agent.invoke({
    "messages": [{"role": "user", "content": "Give me the latest stock news for the ticker symbol APPL(Apple Inc.) including headlines and a short summary of the article"}]
})

```
